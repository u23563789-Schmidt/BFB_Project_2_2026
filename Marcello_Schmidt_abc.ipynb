```python
#Part 2 - ABC Inventory Classification

#This python script classifies a list of SKU's into the ABC inventory tiers and is based on their annual usage
#value, so that the tier A values (high-value) can then be identified. 
```

```python
  skus = [
 {"sku": "BRK-100", "demand": 2000, "cost": 45},
 {"sku": "GSK-220", "demand": 1500, "cost": 30},
 {"sku": "BLT-010", "demand": 10000, "cost": 2},
 {"sku": "BRG-330", "demand": 800, "cost": 60},
 {"sku": "SEAL-500","demand": 3000, "cost": 5},
 {"sku": "MTR-700", "demand": 50, "cost": 800},
 {"sku": "WSH-050", "demand": 20000, "cost": 0.5},
 {"sku": "CBL-900", "demand": 400, "cost": 25},
]
```

```python
def usage_value(demand, cost):
 return demand * cost
for item in skus:
 item["value"] = usage_value(item["demand"], item["cost"])
```

```python
skus_sorted = sorted(skus, key=lambda item: item["value"], reverse=True)
```

```python
total_value = sum(item["value"] for item in skus_sorted)
running_total = 0
for item in skus_sorted:
 running_total += item["value"]
 item["cum_pct"] = (running_total / total_value) * 100
```

```python
def assign_tier(cum_pct):
 if cum_pct <= 80:
     return "A"
 elif cum_pct <= 95:
     return "B"
 else:
     return "C"
for item in skus_sorted:
 item["tier"] = assign_tier(item["cum_pct"])
```

```python
for item in skus_sorted:
 print(item["sku"], "| value:", item["value"],
 "| cum %:", round(item["cum_pct"], 1),
 "| tier:", item["tier"])
```

```python
tier_counts = {"A": 0, "B": 0, "C": 0}
for item in skus_sorted:
 tier_counts[item["tier"]] += 1
print(tier_counts)
```

```python
#Question 2 - Tier counts with 70%/90% thresholds: {'A': 3, 'B': 3, 'C': 4}

def assign_tier_tight(cum_pct):
    if cum_pct<= 70:
        return "A"

    elif cum_pct<= 90:
        return "B"

    else:
        return"C"

for item in skus_sorted:
    item["tier"]=assign_tier_tight(item["cum_pct"])

tier_counts_q2={"A": 0, "B": 0, "C": 0}
for item in skus_sorted:
    tier_counts_q2[item["tier"]] += 1

print("\nQuestion 2")
print("Tier counts with 70%/90% thresholds:", tier_counts_q2)

#By tightening the cuttoffs to 70%/90%, tier A shrinks, tier B stays the same. 
#Therefore more SKU'sare pushed down into tier C.
#Fewer items now qualify in the tier A category.
```
